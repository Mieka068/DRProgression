the union of whatever IS trained (EX + MA); it is not a claim of complete lesion coverage.

## Honest caveats for the presentation

- Segmentation-driven conditioning covers EX + MA only tonight (HE/SE next).
- Module 2's epoch count (5) and optimizer settings here intentionally diverge from
  `train_combined.py`'s existing defaults to match the manuscript's stated Adam
  hyperparameters -- see `train_module2_poc.py`'s module docstring.
- FID/PSNR/SSIM above are from a tiny, short run -- not a claim of matching DRForecastGAN's
  published benchmark (FID 27.3 / PSNR 25.3 / SSIM 0.93).
- Module 2's AdaIN stage-conditioning + autoregressive cascade generation (the manuscript's
  claimed core original contribution) is not implemented yet -- this run uses the existing
  plain conditional GAN in `DRForestGAN-v2/`.


```
# ===== Module 1 segmentation — qualitative masks on FGADR test images =====
# Run AFTER cells 1–6 of this notebook. Does NOT need the training cells.
# Per image:  fundus | EX pred | EX GT | MA pred | MA GT | MA prob-map
import os, glob
import numpy as np, pandas as pd, torch
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
import segmentation_models_pytorch as smp

SEG_DIR    = '/content/drive/MyDrive/Thesis_Datasets/module1_runs/segmentation'
CKPT       = {'EX': f'{SEG_DIR}/models_FGADR_NO_TATL_ex/model_2.pth.tar',
              'MA': f'{SEG_DIR}/models_FGADR_NO_TATL_ma/model_2.pth.tar'}
GT_FOLDER  = {'EX': 'HardExudate_Masks', 'MA': 'Microaneurysms_Masks'}
IMAGE_SIZE = 1280      # matches evaluate_segmentation_dice_iou.py's --image-size; drop to 512 if slow
N_IMAGES   = 6
OUT_PNG    = f'{SEG_DIR}/qual_masks_fgadr.png'

seg_root = '/content/data/FGADR-Seg-set_Release/Seg-set'
if not os.path.isdir(seg_root):
    seg_root = (glob.glob('/content/data/**/Seg-set', recursive=True) or [None])[0]
assert seg_root and os.path.isdir(seg_root), 'FGADR Seg-set not found — run cell 2'
img_dir = os.path.join(seg_root, 'Original_Images')

csv_path = next((p for p in [
    '/content/M2-DRProgression/module1/data/DR_Seg_Grading_Label_Filtered.csv',
    os.path.join(seg_root, 'DR_Seg_Grading_Label_Filtered.csv')] if os.path.isfile(p)), None)
assert csv_path, 'Filtered CSV not found'

names = list(pd.read_csv(csv_path)['Image_Name'])
test_names = names[len(names) - int(len(names) * 0.2):]          # same 60/20/20 split as the eval script
picks = [nm for nm in test_names
         if os.path.isfile(os.path.join(img_dir, nm))
         and all(os.path.isfile(os.path.join(seg_root, GT_FOLDER[l], nm)) for l in ('EX', 'MA'))][:N_IMAGES]
assert picks, 'no test images with both EX and MA GT masks'
print('images:', picks)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
tfm = transforms.Compose([transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)), transforms.ToTensor()])

def load(path):
    m = smp.Unet(encoder_name='resnet50', encoder_weights=None, in_channels=3, classes=2)
    st = torch.load(path, map_location=device)
    m.load_state_dict(st['state_dict'] if isinstance(st, dict) and 'state_dict' in st else st, strict=True)
    return m.eval().to(device)
net = {l: load(p) for l, p in CKPT.items()}

def dice(a, b, e=1e-6):
    a, b = a.astype(bool), b.astype(bool)
    return (2 * (a & b).sum() + e) / (a.sum() + b.sum() + e)

COLS = ['fundus', 'EX pred', 'EX GT', 'MA pred', 'MA GT', 'MA prob']
fig, ax = plt.subplots(len(picks), len(COLS), figsize=(2.6 * len(COLS), 2.6 * len(picks)))
ax = np.atleast_2d(ax)
agg = {'EX': [], 'MA': []}

for r, name in enumerate(picks):
    img = Image.open(os.path.join(img_dir, name)).convert('RGB'); W, H = img.size
    x = tfm(img).unsqueeze(0).to(device)
    cells, labels = [np.asarray(img)], ['']
    with torch.no_grad():
        for l in ('EX', 'MA'):
            prob = torch.softmax(net[l](x), 1)[0, 1].cpu().numpy()
            prob = np.asarray(Image.fromarray((prob * 255).astype('uint8')).resize((W, H))) / 255.0
            pred = (prob > 0.5).astype('uint8')
            gt = (np.asarray(Image.open(os.path.join(seg_root, GT_FOLDER[l], name)).convert('L')) > 127).astype('uint8')
            d = dice(pred, gt); agg[l].append(d)
            cells += [pred, gt]; labels += [f'Dice {d:.3f}', f'{int(gt.sum())} GT px']
            if l == 'MA':
                cells.append(prob); labels.append(f'max p = {prob.max():.2f}')
    for c in range(len(COLS)):
        a, im = ax[r, c], cells[c]
        a.imshow(im, cmap=None if im.ndim == 3 else ('magma' if COLS[c] == 'MA prob' else 'gray'),
                 vmin=None if im.ndim == 3 else 0, vmax=None if im.ndim == 3 else 1)
        if r == 0: a.set_title(COLS[c], fontsize=10)
        if c == 0: a.set_ylabel(name, fontsize=9)
        if labels[c]: a.set_xlabel(labels[c], fontsize=8)
        a.set_xticks([]); a.set_yticks([])

plt.tight_layout(); plt.savefig(OUT_PNG, dpi=110, bbox_inches='tight'); plt.show()
print(f'\nsaved -> {OUT_PNG}')
print(f'mean Dice over these {len(picks)} images:  EX {np.mean(agg["EX"]):.3f}   MA {np.mean(agg["MA"]):.3f}')
```
